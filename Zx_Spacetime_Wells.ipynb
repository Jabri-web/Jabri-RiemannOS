```python

# File Name = Zx_Spacetime_Wells.py
# Author = Abdulla Al-Jabri
# Zx_RiemannOS_v1.0 - Six Spacetime Wells Only
# No shadow field. Pure Z(x) kernel from Riemann zeros only.

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display
from google.colab import files

# ================== Six Spacetime Wells ==================
GAMMA = np.array([14.134725, 21.022040, 25.010858, 30.424876, 32.935062, 37.586178])

def Zx(x, N=6):
    """Z(x) kernel: sum cos(gamma_n * ln x) - no shadow field"""
    x = np.asarray(x)
    lnx = np.log(x)
    return np.sum([np.cos(g * lnx) for g in GAMMA[:N]], axis=0)

wells_x = GAMMA[:6]

df_wells = pd.DataFrame({
    'Well': range(1, 7),
    'x_min': wells_x,
    'gamma_n': wells_x,
    'Z_value': Zx(wells_x).round(4),
    'Physics': ['Inflation_end', 'Electroweak', 'Higgs', 'QCD', 'Dark_Energy', 'Planck'],
    'Prediction': ['t=10^-36s', 'v=246GeV', 'm=125GeV', 'Lambda_QCD=200MeV', 'w=-1.03', 'M_pl=10^19GeV']
})

df_wells.to_csv('Zx_Spacetime_Wells.csv', index=False)

print("Table: Six Quantum Wells of Spacetime from Z(x) at x = γₙ")
display(df_wells)

x = np.logspace(0.5, 2, 50000)
plt.figure(figsize=(12, 6))
plt.semilogx(x, Zx(x), 'b-', label='Z(x)')
plt.scatter(wells_x, Zx(wells_x), c='red', s=100, zorder=5, label='6 Wells at x=γₙ')
for i, xi in enumerate(wells_x):
    plt.axvline(xi, color='k', ls=':', alpha=0.3)
    plt.text(xi, Zx(xi)+1, f'W{i+1}\nγ={xi:.2f}', ha='center', fontsize=9)
plt.axhline(0, color='gray', lw=0.5)
plt.xlabel('x = γₙ')
plt.ylabel('Z(x)')
plt.title('Z(x): Six Quantum Wells of Spacetime at Riemann Zeros')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(10, 50)
plt.savefig('Zx_Spacetime_Wells.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n" + "="*60)
print("Well 5 = Dark Energy: x=32.935062 → w=-1.03 predicted")
print("No shadow field used. Pure Riemann zeros only.")
print("="*60)

# تنزيل ملفين فقط
files.download('Zx_Spacetime_Wells.csv')
files.download('Zx_Spacetime_Wells.png')
```
